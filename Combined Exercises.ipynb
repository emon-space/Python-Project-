```python
# Q10. [MODULES 1 + 2 + 3 COMBINED - Full Exam-Style Problem]
#
# Objective:
# A weather station logs daily temperatures as ONE raw string of
# space-separated integers (which may be negative), e.g.:
#   "5 -2 8 0 -3 6"
#
# Write a function weather_report(raw_data, filename) that:
#
#   1. Parses raw_data into a list of integer temperatures.
#
#   2. Prints the running average temperature after each day is added,
#      in the format "Day <n>: <average>" (rounded to 2 decimals) --
#      exactly like an accumulator problem from class.
#
#   3. Writes a horizontal bar chart to a text file called `filename`,
#      one line per day: a '|' marks zero, with '=' characters
#      extending RIGHT for positive temperatures and LEFT for negative
#      temperatures (identical convention to the bar chart from class).
#
#   4. Returns the list of parsed integer temperatures.
#
# Example:
# >>> weather_report("5 -2 8 0 -3", "report.txt")
# Day 1: 5.0
# Day 2: 1.5
# Day 3: 3.67
# Day 4: 2.75
# Day 5: 1.6
# [5, -2, 8, 0, -3]
#
# (and report.txt contains, one line per day:)
# |=====
# ==|
# |========
# |
# ===|
```

```python
def weather_report(raw_data, filename):

    string_pieces = raw_data.split()
    temp = [] # -> [5, -2, 8, 0, -3]
    for t in string_pieces:
        temp.append(int(t))


    total = 0
    count = 0

    for t in temp: # -> [5, -2, 8, 0, -3]
        count += 1
        total += t

        average = total/count
        print(f'Day {count}  : {round(average,2)}')

    with open(filename,'w') as f:
        for t in temp:
            if t>0:
                bar = "|"+ ('='*t)
            elif t<0:
                bar = ('='*abs(t)) + '|'
            else:
                bar = '|'

            f.write(bar + "\n")


weather_report("5 -2 8 0 -3", "report.txt")

```

```python
# Q10. [MODULES 4 + 5 + 6 COMBINED - Full Exam-Style Problem]
#
# Objective:
# A gym's file contains two sections separated by a BLANK line.
#
# Section 1 -- class prices, one per line, TAB-separated:
#   <ClassName TAB Price>       (ClassName may contain spaces)
#
# Section 2 -- bookings, one per line, TAB-separated, THREE fields:
#   <MemberID TAB MemberName TAB ClassesCommaSeparated>
#   (a class may appear more than once in one member's booking list,
#   meaning they attended it multiple times -- count it each time)
#
# Write a function gym_report(filename) that:
#   1. Reads and splits the file into the two sections.
#   2. Builds a dictionary mapping class name (lowercase) -> price.
#   3. For each booking, computes the total cost (skip any class name
#      not found in the price dictionary -- do not crash).
#   4. Returns a list of (total_cost, member_id, member_name) tuples,
#      SORTED by total_cost DECREASING, then by member_id INCREASING.
#   5. Also writes each line to a file "gym_bills.txt" in the format:
#      "<total_cost>\t<member_id>\t<member_name>"
#
# Example: (file "gym.txt" contains:)
#   Yoga	10
#   Spinning	15
#   Pilates	12
#
#   3	Elena	Yoga,Spinning
#   1	Marco	Pilates
#   5	Luca	Yoga,Yoga,Pilates
#
# >>> gym_report("gym.txt")
# [(32, 5, 'Luca'), (25, 3, 'Elena'), (12, 1, 'Marco')]
#
# (and gym_bills.txt contains:)
# 32	5	Luca
# 25	3	Elena
# 12	1	Marco
```

```python
def  gym_report(filename):
    with open(filename,'r') as file:
        content = file.read().strip()

    sections = content.split('\n\n')
    section1 = sections[0] #Prices
    section2 = sections[1] #Booking

    price_dict = {} #-> {'yoga': 10, 'spinning': 15, 'pilates': 12}

    for line in section1.split("\n"):
        if line.strip("\t"):
            name,price = line.split("\t")
            price_dict[name.lower()] = int(price)


    results = []
    for line in section2.split("\n"):
        if line.strip():
            m_id,m_name,class_str = line.split("\t")

            total_cost = 0

            for c in class_str.split(','):
                clean_class = c.strip().lower()

                if clean_class in price_dict:
                    total_cost += price_dict[clean_class]

            results.append(total_cost,int(m_id),m_name)


    print(results)



gym_report('gym.txt')
```
