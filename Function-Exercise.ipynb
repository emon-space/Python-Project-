```python
#Solution
def print_right(text):
    space_needed = 40 -len(text)
    print(" "*space_needed+text)

# Test it
print_right("Monty")
print_right("Python's")  
print_right("Flying Circus")
```

```python
# Practice Problem (Optional)
def print_center(text):
    total_space = 40 - len(text)
    space_need = total_space//2
    print(" "* space_need +text)
    
print_center('Monty')
```

```python
#Solution
def triangle(string,height):
    for i in range(1,height+1):
        print(string*i)

triangle('T',5)
    
```

```python
#Practice Problem (Optional)
def reverse_triangle(text,height):
    for i in range(height,0,-1): #(start,stop,steps)
        print(text*i)

reverse_triangle('L',5)
```

```python
#Solution
def rectangle(string,width,height):
    for i in range(height):
        print(string*width)
    
rectangle('H',5,4)
```

```python
#Practice Problem(Optional)
def hollow_rectangle(text,width,height):
    for i in range(height):
        if i == 0 or i == height-1:
            print(text*width)
        else:
            print(text+" "*(width-2)+text)
            
hollow_rectangle('*', 6, 4)

```

```python
#Solution
def bottle_verse(number):
    #1st Line 
    if number == 1:
        print(f'{number} bottle of beer on the wall')
    else:
        print(f'{number} bottle of beer on the wall')

    #2nd Line
    if number == 1:
        print(f'{number} bottle of beer')
    else:
        print(f'{number} bottle of beer ')
    
    #3rd Line 
    print('Take one down, pass it around')

    # 4th Line 
    next_number = number - 1
    if next_number == 1:
        print(f"{next_number} bottle of beer on the wall")
    elif next_number == 0:
        print(f"{next_number} bottles of beer on the wall")
    else:
        print(f"{next_number} bottles of beer on the wall")

# Test it
bottle_verse(99)
print()
bottle_verse(3)
print()
bottle_verse(1)


```
