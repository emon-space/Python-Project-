```python
def parse_bar_chart(filename):
    values = [ ]

    with open(filename,'r') as file:
        for line in file:
            left_side , right_side= line.split('|')
            positive_value = right_side.count('=')
            negative_value = left_side.count('=')
            final_value = positive_value-negative_value
            values.append(final_value)
        print(values)


parse_bar_chart('chart.txt')
```

```python
#Zig_Zag Patten
def zigzag(width,height):
    positon = 0
    direction = 1

    for i in range(height):
        line = ["-"]*width
        line[positon] ="*"
        print(''.join(line))
        positon += direction

        if positon == width -1:
            direction = -1
        elif positon == 0:
            direction =1


zigzag(5,9)
```

```python
# Sawtoodh Pattern
def sawtooth(width,heihgt):
    position = 0

    for i in range(heihgt):
        line = ['-']*width
        line[position] = "*"
        print(''.join(line))
        position += 1

        if position == width :
            position = 0





sawtooth(5,7)
```

```python
# Circle
import math
def  circle(W,H):
    radius = min(W,H)/2
    center_x = (W-1)/2
    center_y = (H-1)/2

    for y in range(H):
        row = " "
        for x in range(W):
            distance = math.sqrt((x - center_x)**2 +(y- center_y)**2)

            if distance <= radius:
                row+= '*'
            else:
                row+= '-'
        print(row)
circle(7,5)
```

```python
# Intergral of linear Piece Wise Function
def compute_intergral(filename):
    with open(filename,'r')as file:
        points = []
        for line in file:
            line = line.split()
            x,y = map(float,line)
            points.append((x,y))

        total_area = 0
        for i in range(len(points)-1):
                x1,y1  = points[i]
                x2,y2 = points[i+1]
                area = (x2-x1)*(y1+y2)/2
                total_area += area
        print(f'The intergral of function: {total_area:.1f}')




compute_intergral('points.txt')
```
