### Accumalator Pattern Problem:

Leibniz formula:




```python

def approximate_pi(N):
    total = 0
    for n in range(N+1):
        term = ((-1)**n) / (2*n + 1)
        total = total+term
    pi = 4*total
    return pi

approximate_pi(10)
```

```python
def log_approximate(x,num_terms):
    total = 0
    for n in range(1,num_terms+1):
        numerator = ((-1)**(n+1))*(x-1)**n
        denumarator = n
        term = numerator/denumarator
        total = total+term
        return total
log_approximate(1.5,5)

```

```python
import math
def approximate_e(N):
    total = 0
    for n in range(N+1):
        terms = 1/math.factorial(n)
        total = total = terms
    return total
approximate_e(5)
```

```python
def geomatric_sum(r,N):
    total = 0
    for n in range(N+1):
        term = r**n
        total = total +term
    return total

geomatric_sum(5,3)
```

```python
import math
def sin_approx(x,N):
    total = 0
    for n in range(0,N+1):
        numerator = ((-1)**n)*(x**(2*n+1))
        denumerator = math.factorial(2*n +1)
        term = numerator/denumerator
        total += term
    return total
sin_approx(4,3)
```

### Practice Problem



```python
# Q1. [MODULE 1 - Series Approximation]
#
# Objective:
# The number e can be approximated using the series:
#   e ≈ Σ (n=0 to N) 1/n!
# Write a function e_approx(num_terms) that computes this approximation
# using num_terms terms (n = 0, 1, ..., num_terms-1), printing the partial
# sum after EACH term is added.
#
# Example:
# >>> e_approx(4)
# 1.0
# 2.0
# 2.5
# 2.6666666666666665
```

```python
def e_approx(num_terms):
    import math
    total = 0

    for n in range(num_terms):
        e = 1/math.factorial(n)
        total += e
        print(total)

e_approx(4)
```

```python
# Q2. [MODULE 1 - Alternating Series with a Variable Argument]
#
# Objective:
# The arctangent function can be approximated for -1 <= x <= 1 using:
#   arctan(x) ≈ Σ (n=0 to N) [(-1)^n * x^(2n+1)] / (2n+1)
# Write a function arctan_approx(x, num_terms) that computes the
# approximation using num_terms terms (n = 0 ... num_terms-1),
# printing the partial sum after each term.
#
# Example:
# >>> arctan_approx(1, 4)
# 1.0
# 0.6666666666666667
# 0.8666666666666667
# 0.7238095238095239
```

```python
def arctan_approx(x, num_terms):
    total = 0
    for n in range(num_terms):
        numerator = ((-1)**n) * (x**(2*n+1))
        denumerator = 2*n +1
        arctan = numerator /denumerator
        total += arctan
        print(total)
arctan_approx(1,4)
```

```python
# Q3. [MODULE 1 - Exact Output Format Discipline]
#
# Objective:
# Write a function harmonic_partial(num_terms) that computes the partial
# sums of the harmonic series:
#   H_n = Σ (k=1 to n) 1/k
# For each n from 1 to num_terms, print the current partial sum and the
# term index n, separated by a TAB character, in the exact format:
#   "<partial_sum> \t <n>"
#
# Example:
# >>> harmonic_partial(4)
# 1.0 	 1
# 1.5 	 2
# 1.8333333333333333 	 3
# 2.083333333333333 	 4
```

```python
def harmonic_partial(num_terms):
    total = 0
    for n in range(1,num_terms+1):
        Hn = 1/n
        total += Hn
        print(total,'\t',n) # n -> iteration number

harmonic_partial(4)
```
