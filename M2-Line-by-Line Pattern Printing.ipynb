## Line-by-Line Pattern Printing Problems

```python
def zigzag(weight,height):
    position  = 0
    direction = 1

    for _ in range(height):
        line = ["-"]* weight
        line[position] = "*"
        print("".join(line))
        position += direction

        if position  == weight -1:
            direction = -1
        elif position == 0:
            direction = 1

zigzag(5,12)

```

```python
def sawtooth(W,H):
    position = 0
    for _ in range(H):
        line = ['-']*W
        line[position] = '*'
        print(''.join(line))
        position = (position+1)%W
sawtooth(5,12)
```

```python
# Objective:
# Write a function named circle that prints an approximated circular pattern inside a box of given dimensions.

# Details:
# 	•	The pattern is centered within a box of width W and height H.
# 	•	The circle is drawn using asterisks (*), and the empty space is filled with minus signs (-).
# 	•	The center of the circle is at ((H-1)/2, (W-1)/2).
# 	•	The ordinate axis is reversed, meaning the top row corresponds to the highest y-coordinate.

# Example Output:
# For circle(7, 5), the output is:

# --***--
# -*****-
# -*****-
# -*****-
# --***--
```

```python
import math
def circle(W,H):
    radius = min(W,H)/2
    center_x = (W-1)/2
    center_y = (H-1)/2

    for y in range(H):
        row = ''
        for x in range(W):
            distance = math.sqrt((x - center_x)**2 + (y-center_y)**2)

            if distance <= radius:
                row = row +'*'
            else:
                row = row + '-'
        print(row)


circle(7,5)
```

```python
def bar_chart(values,filename):
    with open(filename,'w') as f:
        for v in values:
            bar = "="*abs(v)

            if v>= 0:
                line = '|'+ bar
            else:
                line = bar + '|'
            f.write(line + '\n')

bar_chart([5,-2,8],'chart.txt')
```

```python
# Q4. [MODULE 2 - Geometric-Test Family]
#
# Objective:
# Write a function diamond(W, H) that prints a W x H grid where a cell
# is '*' if the sum of the absolute horizontal and vertical distances
# from the grid's center (Manhattan distance, NOT Euclidean) is less
# than or equal to a radius = min(W, H) // 2, and '-' otherwise.
#
# Example:
# >>> diamond(7, 5)
# ---*---
# --*-*--
# -*---*-
# --*-*--
# ---*---
```

```python
def diamond(W, H):
    radius = min(W,H)//2
    center_x = W // 2
    center_y = H//2


    for y in range(H):
        row = ""
        for x in range(W):
            dx = abs(x-center_x)
            dy = abs(y - center_y)
            distance = dx+dy

            if distance <= radius:
                row += '*'
            else:
                row += '-'

        print(row)

diamond(7, 5)
```

```python
# Q5. [MODULE 2 - Value-Driven Family, Vertical Orientation]
#
# Objective:
# Write a function histogram(values) that prints a VERTICAL bar chart
# (unlike the horizontal one from class). Each value is a non-negative
# integer; each bar is one column wide, made of '*', growing UPWARD from
# the bottom. Bars are printed side by side, separated by a single space,
# padded with blank spaces above shorter bars so all columns line up.
#
# Example:
# >>> histogram([3, 1, 2])
#     *
#     *
# *   *   *
# *   *   *
# *   *   *
```

```python
def histogram(values):
    if not values:
        return
    max_height = max(values)

    for level in range(max_height,0,-1):
        row_chart = []
        for val in values:
            if val <= level:
                row_chart.append("*")
            else:
                row_chart.append('')

        row_string = ''.join(row_chart)
        print(row_string)


histogram([3, 1, 2])
```
