```python
#Exam Question:
# Objective:
# You have a menu with food items and their prices, separated by a tab. Customers at different tables place orders, specifying their person ID, table ID, and the items they want (comma or space-separated). The goal is to calculate the total cost for each person and print the result in the format:

# Total Cost | Person ID | Table ID

# The output must be sorted by cost (ascending), then by person ID, then by table ID.

# Example Input:

# Menu:
# Patate   3
# Carote   2
# Polenta  4

# Orders:
# 1 2 patate, polenta, carote
# 2 3 patate polenta

# Example Output:
# 7 2 3
# 20 1 2
```

```python
# Testing data
menu_file = [
    "Patate 3",
    "Carote 2",
    "Polenta 4"
]
order_file = [
    "1,2,Patate,Carote,Polenta",
    "2,3,Patate,Polenta"
]
```

```python
#Create the Menu
menu = {}
for line in menu_file:
    food,price = line.split()
    menu[food] = int(price)
    print(menu)
```

```python
#calculating each order
result = []
for line in order_file:
    parts = line.split(',')

    person_ID = int(parts[0])
    table_ID = int(parts[1])
    foods = parts[2:]

    #Tracking total
    total = 0
    for food in foods:
        total = total + menu[food]
    result.append((total,person_ID,table_ID))
#Sorting and print
result.sort()
for total,person_ID,table_ID in result:
    print(total,person_ID,table_ID)
```

```python
# Q1. [MODULE 3 - Cleaning and Reformatting Raw Text]
#
# Objective:
# You are given raw configuration text, one "key = value" pair per line,
# with inconsistent spacing around the '=' sign. Write a function
# clean_config(raw_lines) that returns a list of strings, each formatted
# EXACTLY as "Key: Value" (key capitalized, single colon-space separator,
# no extra whitespace), preserving the original line order.
#
# Example:
# >>> raw_lines = ["name =John", "age= 23", "city   =   Milan"]
# >>> clean_config(raw_lines)
# ['Name: John', 'Age: 23', 'City: Milan']
```

```python
def clean_config(raw_lines):
   clean_line = []
   for line in raw_lines:
       key,value = line.split("=")

       clean_key = key.strip()
       clean_value = value.strip()

       formated_line = f"{clean_key.capitalize()}: {clean_value}"

       clean_line.append(formated_line)
   return clean_line

clean_config(raw_lines=["name =John", "age= 23", "city   =   Milan"])
```

```python
# Q2. [MODULE 3 - Slicing for Fixed-Format Text]
#
# Objective:
# A student ID code is always 11 characters long, structured as:
#   first 2 characters = faculty code
#   next 7 characters  = registration number
#   last 2 characters  = enrollment year (2 digits)
# Write a function format_id(raw_id) that takes such a code (e.g.
# "AI0451231" -- wait, use the real length) and returns a formatted
# string "FACULTY-NUMBER-20YEAR" using slicing (no split()).
#
# Example:
# >>> format_id("AI045123124")
# 'AI-0451231-2024'
```

```python
def format_id(raw_id):
    faculty = raw_id[0:2]
    registration_number = raw_id[2:9]
    enrollment_year = raw_id[9:]

    formated_string = f'{faculty}-{registration_number}-20{enrollment_year}'
    return formated_string

print(format_id("AI045123124"))

```

```python
def running_average(raw_line):
    string_list = raw_line.split(",")
    total_sum = 0
    day_count = 0

    for item in string_list:
        curren_number = int(item)
        total_sum += curren_number
        day_count += 1
        average = round(total_sum/day_count,2)
        print(f'Day{day_count} :  {total_sum}')

raw_line="3,7,2,9,5"
running_average(raw_line)
```
