```python
#Tool 2 — Flattening: when a comprehension has two fors
with open('food.txt','r') as file:
    items = []
    for line in file:
        words = line.strip().split()
        for word in words:
            items.append(word)
    print(items)
```

```python
#Integral of a Piecewise Linear Function

# Objective:
# Write a function to compute the integral of a piecewise linear function described by a set of points (knots).
#
# Details:
# 	•	The function is defined between x_1 and x_N using linear functions connecting each pair of consecutive points.
# 	•	The integral is calculated as the sum of the areas of trapezoids formed by these points.
# 	•	Input is a text file where each line has the x and y coordinates of a point.
# 	•	The points are listed in ascending order of x.
#
# Example Input:
#
# 0 0
# 1 2
# 3 1
# 4 3

# Example Output:
# The integral of the function described by the knots is 5.0
```

```python
def compute_intergral(file_name):
    with open(file_name,'r') as file:
        points = []
        for line in file:
            X,Y = map(float,line.split())
            points.append((X,Y))


    total_area = 0
    for i in range(len(points) -1):
        X1,Y1 = points[i]
        X2,Y2 = points[i+1]

        trapoziod_area = (X2-X1)*(Y1+Y2)/2
        total_area += trapoziod_area
    print(f"The Intergral of the function described by the knots is {total_area:.1f}")

compute_intergral('points.txt')
```

```python
# Q1. [MODULE 4 - Stride Slicing / Un-interleaving]
#
# Objective:
# A sensor logs data as one flat list where readings alternate between
# TEMPERATURE and HUMIDITY: [temp1, hum1, temp2, hum2, ...].
# Write a function split_readings(flat_list) that returns a tuple of
# two lists (temperatures, humidities), using slicing with a stride
# (no manual loop with an if-condition allowed).
#
# Example:
# >>> split_readings([21, 60, 23, 58, 19, 65])
# ([21, 23, 19], [60, 58, 65])
```

```python
def split_readings(flat_list):
    temperature = flat_list[0::2]
    huidities = flat_list[1::2]
    return (temperature,huidities)

print(split_readings([21, 60, 23, 58, 19, 65]))
```
