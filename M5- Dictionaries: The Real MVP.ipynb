```python
# Tool 2 — Building a dictionary from a loop: turning a list into a lookup table
menu_lines = [
    "Burger 10",
    "Pizza 15",
    "Fries 5"
]
def parse_menu(menu_lines):
    menu ={}
    for line in menu_lines:
        item,price = line.strip().split()
        menu[item.lower()] = int(price)
    return menu



parse_menu(menu_lines)
```

```python
# Tool 3 — Safe lookup: .get() and the in keyword
menu = {
    "patate": 3,
    "carote": 2
}
price = menu.get('carote',0)
print(price)
```

```python
#Pseudosymmetrical Strings

# objective:
# Consider only UPPERCASED characters and numbers.
# On the keyboard there are many characters who are not exactly the same, but seem symmetrical to each other, like

# L and J
# Z and 5
# 1 and I
# …

# A string in pseudosymmetrical if the first and the last, the second and the second to last, etc. characters are pseudosymmetrical.
# If a string has an odd number of characters, the central one should also be pseudosymmetrical to itself.

# Write a function that checks if a set of strings given line per line in a file are Pseudosymmetrical and print them in alphabetical order. The first line in the file declares what are the pseudosymmetrical characters, given in couples separated by a space, like:

# LJ O0 XX …

```

```python
#Just Build the rule
def build_rules(line):
    rules = {}
    for pair in line.split():
        left = pair[0]
        right = pair[-1]

        rules.setdefault(left,set()).add(right)
        rules.setdefault(right,set()).add(left)
    return rules

# Check if one word is pseudosymmetrical
def is_valid(word,rules):
    left = 0
    right = len(word)-1

    while left <= right:
        if word[left] not in rules:
            return False

        if word[right] not in rules[word[left]]:
            return False
        left += 1
        right -= 1
    return True

# Read the file and print valid words
def check_file(filename):
    with open(filename) as file:
        rules = build_rules(file.readline())

        valid_words = []

        for line in file:
            word = line.strip()

            if is_valid(word,rules):
                valid_words.append(word)

        for word in sorted(valid_words):
            print(word)

check_file('input.txt')

```
