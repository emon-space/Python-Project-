## File I/O & Parsing Real-World Messy Text

```python
#Tool 1 — Opening a file, and why we always write with
with open('input.txt','r') as f:
    content = f.read()
print(content)
```

```python
#Tool 2 — Three ways to read a file, and when each one is the right tool

# Option A — one line at a time, memory-efficient, the default choice
with open('input.txt') as f:
    for line in f:
        print(line.strip())

# Option B — all lines at once, as a list of strings (with \n still attached)
with open('input.txt') as f:
    lines = f.readlines()
    print(line)

# Option C — the entire file as ONE single string
with open('input.txt') as f:
    content = f.read()
    print(content)
```

```python
#Tool 3 — Detecting where one section ends and another begins: a new kind of "state"
def seperate_section (filename ):
    menu = []
    order = []
    is_menu = True

    with open (filename) as file:
        for line in file:
            if line.strip() == "":
                is_menu = False
                continue

            if is_menu:
                menu.append(line.strip())
            else:
                order.append(line.strip())

    return menu, order

menu, order = seperate_section('food.txt')
print(menu)
print(order)




```
