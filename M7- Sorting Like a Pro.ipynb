```python
#Tool 4 — Mixing ascending and descending: the negation trick
orders = [(20, 3, 2), (20, 1, 5), (7, 5, 1)]
result = sorted(orders,key=lambda o:(-o[0],o[1],o[2]))
print(result)
```

```python
#Tool 5 — Sorting custom objects: teaching Python how to compare your things
class Die: # Class
    def __init__(self,faces,val): #Method
        self.faces = faces
        self.val = val

    def __repr__(self):  #Method
        return f'D {self.faces} {self.val}'

    def __lt__(self, other): #Method
        if self.val == other.val :
            return self.faces > other.faces
        else:
            return self.val < other.val


d1 = Die(6,4)
d2 = Die(4,4)
d3 = Die(20,2)

Dice  = [d1,d2,d3]
print('Before:',Dice)
Dice.sort(reverse=True)
print("After:", Dice)



```
