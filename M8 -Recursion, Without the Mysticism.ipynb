```python
#Tool 1 — The simplest possible recursive function, traced by hand
def count_down(n):
    if n == 0:
        print('Liftoff!')
        return
    print(n)
    count_down(n-1)
count_down(3)
```

```python
#Tool 2 — Recursion that returns and combines: the part that actually confuses people
def factorial(n):
    if n == 0:
        return 1
    return  n*factorial(n-1)
factorial(3)
```

```python
#q8 pseudo()
mapping = {
    "L": {"J"},
    "J": {"L"},
    "O": {"O"}
}

def pseudo(string,mapping):
    if string[0] in mapping[string[-1]]:
        return True
    else:
        return False

print(pseudo("LOOJ", mapping))


```
