```python
# Tool 3 — __eq__ and __lt__: teaching == and < what "equal" and "less than" mean for your fobject
class Die:
    def __init__(self,faces,val): #Method -1
        self.faces = faces
        self.val = val

    def __str__(self): #Method-2
        return f'D {self.faces} ({self.val})'

    def __eq__(self, other):#Method-3
        return self.val == other.val and self.faces == other.faces

    def __lt__(self, other):#Method-4
        if self.val == other.val:
            return self.faces > other.faces
        return self.val < other.val

a = Die(6,4)
b = Die(4,4)
c = Die(6,4)

print(a == c)
print( a == b)
print(a < c)
print(c < a)




```

```python
#Short Experiment:Goal:Create a class called MutableStr that behaves like a string, but lets the user change individual characters
#example :
#s = MutableStr('hello')
# s[0]  = "H"
# print(s)
```

```python
class MutalbleStr(list):
    def __int__(self,inputString):
        super().__int__(inputString)
s = MutalbleStr("hello")
s[0] = "H"
s.append('!')

print(s)
```

```python
#Tool 5 — __add__: teaching the + symbol new behavior
class MutableStr(list):
    def __add__(self, other):
        left = list(self)
        right = [c for c in other]
        combinded = left+right
        return MutalbleStr(combinded)

s = MutableStr("hi")

t = s + "there"
print(t)
```

```python

# q10  MutableStr

# Objective: in Python strings are immutable objects; which means that you cannot reassign a
# character of the string like:

# str1 = “hello”
# str1[2] = “a”

# Write a class in python, who inherits its behaviour from the list class and allows to create objects MutableStr
# in order to execute the following code:

# s = MutableStr(“hello”)
# s.append(“, how are you?”)
# s.change(5, “!”)
# s.change(5, “!!!!!”, 5)
# t = s + “hi”
# print(t)

# where the function change has two ways of working:

# 1. only two parameters, respectively index and string; in this case it replace the character at the specified index
# (if it exists) with the first character of the string passed.

# 2. ⁠3 parameters, respectively index, string, number of characters to change; in this case if there are enough
# characters to be changed, the function starts replacing characters from the specified index until the given
# number of characters is replaced by each of the characters of the passed string until possible
```

```python
class MutableStr(list):
    def __init__(self,inputString):
        super().__init__(inputString)

    def __str__(self):
        return "".join(self)

    def append(self,text):
        for ch in text:
            super().append(ch)

    def change(self,index,text,n=None):
        #verion-1: only change one charecter
        if n is None:
            if index < len(self):
                self[index] = text[0]
        #version-2:replace n characters
        else:
            for i in range(n):
                if index+i >= len(self):
                    break
                if i >= len(text):
                    break
                self[index+i] = text[i]

s = MutableStr('hello')
s.append('!!')
s.change(5,"!")
s.change(5,"ABCDE",5)
print(s)
```
