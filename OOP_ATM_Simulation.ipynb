### How to Use This ATM Simulation

Welcome to this simple ATM simulation! Here's how you can interact with it:

1.  **Run the ATM Class Definition:** First, ensure that the code cell containing the `Atm` class definition (the main blueprint of the ATM) is executed. This makes the `Atm` class available for use.
2.  **Start the ATM:** Next, run the code cell below, `my_atm = Atm()`. This creates an instance of the ATM, and its menu will automatically appear in the output.
3.  **Follow the Menu Prompts:** The ATM will then guide you through its options. You can:
    *   **Create a PIN** (select `1`) to secure your account.
    *   **Deposit money** (select `2`) into your account using your PIN.
    *   **Withdraw money** (select `3`) from your account, also requiring your PIN.
    *   **Check your balance** (select `4`) to see your current funds.
    *   **Exit** (select `5`) when you are done.

Experiment with different scenarios, like trying to withdraw more than you have or using an incorrect PIN, to see how the ATM responds!

```python
class Atm:
  # __init__ is a constractor
  def __init__(self):
    self.pin = ''
    self.blance = 0
    self.menu()

  def menu(self):
    user_input = input("""
    Hello, how would you like to proceed?
    1. Enter 1 to create pin
    2. Enterr 2 to deposit
    3.Enter 3 to withdraw
    4.Enter 4 to check blance
    5.Enter 5 to exit
    """)

    if user_input == '1':
      self.create_pin()
    elif user_input == '2':
      self.deposit()
    elif user_input == '3':
      self.withdraw()
    elif user_input == '4':
      self.check_blance()
    else:
      print("Bye")

  # Create method for each funtionality
  def create_pin(self):
    self.pin = input('Enter Your Pin')
    print("Pin set successfully ")

  def deposit(self):
    temp = input("Enter your pin")
    if temp == self.pin:
      amount = int(input("Enter the amount"))
      self.blance = self.blance + amount
      print("Deposite Succesfully")
    else:
      print("Pin is invalid")

  def withdraw(self):
    temp = input("Enter your pin")
    if temp == self.pin:
      amount = int(input("Enter the amount"))
      if amount < self.blance:
        self.blance = self.blance - amount
        print('Operation Successful')
      else:
        print("Insufficiant Balance")
    else:
      print('Invalid Pin')

  def check_blance(self):
    temp = input("Enter your pin")
    if temp == self.pin:
      print(self.blance)
    else:
      print('Invalid Pin')
```

```python
my_atm = Atm()
```
